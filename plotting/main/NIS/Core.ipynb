```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
# Load the uploaded Excel files
mets_df = pd.read_excel('<EXTERNAL_DATA_ROOT>')
rxns_df = pd.read_excel('<EXTERNAL_DATA_ROOT>')

# Extract model names from the 'Row' column and shorten them
mets_df['Row'] = mets_df['Row'].str.extract(r'Thermo(\w+)PTYGVitconstrained')
rxns_df['Row'] = rxns_df['Row'].str.extract(r'Thermo(\w+)PTYGVitconstrained')

# Rename columns similarly
mets_df.columns = ['Row'] + [col.replace('Thermo', '').replace('PTYGVitconstrained', '') for col in mets_df.columns[1:]]
rxns_df.columns = ['Row'] + [col.replace('Thermo', '').replace('PTYGVitconstrained', '') for col in rxns_df.columns[1:]]

# Set 'Row' as index for heatmap plotting
mets_df.set_index('Row', inplace=True)
rxns_df.set_index('Row', inplace=True)

# Plot heatmaps with model names 
plt.figure(figsize=(14, 6))

# Mets similarity heatmap
sns.clustermap(mets_df, cmap='Oranges', method='average',vmin=0, vmax=1)

sns.clustermap(rxns_df, cmap='Blues',method='average',vmin=0, vmax=1)
plt.show()
```
